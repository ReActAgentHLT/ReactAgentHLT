# Datasets Fetch and Preprocessing

To fetch the 2 datasets is possible to use the 2 dedicated functions `hotpotqa`
and `gsm8k` as a wrapper to HuggingFace API.

Those functions have some additional checks and for GSM8K dataset a **parsing**
utility that cleans the `answer` field.


```python
from datasets import load_dataset

from reactagenthlt.preprocessing import gsm8k, hotpotqa
```

## HotPotQA

For HotPot is possible to decide between `distractor` and `fullwiki` setting and
between `train` and `validation` split.


```python
qa = hotpotqa(setting="distractor", train=False)
qa
```

## GSM8K

For GSM8K be careful on the `reasoning` parameter since it determines if the
`answer` field will contains also the reasoning steps to form expressions. For
evaluation only the final numerical result is needed.


```python
gsm = gsm8k(setting="main", train=False, reasoning=True)
gsm["answer"].iloc[0]
```

```python
gsm = gsm8k(setting="main", train=False, reasoning=False)
gsm["answer"].iloc[0]
```

### Comma Parsing

Another thing to be aware of is that GSM8K has this convention in which a
integer has commas every thousand.

So in the dataset is possible to find `1,234` but is actually `1234`. The
`gsm8k` function just normalizes each of those cases by removing commas.


```python
print(load_dataset("openai/gsm8k", "main")["test"].to_pandas()["answer"].iloc[505])
```

```python
print(gsm["answer"].iloc[505])
```
